```python
from collections import deque
import matplotlib.pyplot as plt
import networkx as nx

# Adjacency List
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# BFS
def bfs(start):
    visited, queue, order = set(), deque([start]), []

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            order.append(node)
            queue.extend(graph[node])

    return order

# DFS
def dfs(node, visited=None, order=None):
    if visited is None:
        visited, order = set(), []

    visited.add(node)
    order.append(node)

    for neighbour in graph[node]:
        if neighbour not in visited:
            dfs(neighbour, visited, order)

    return order

# Traversals
bfs_order = bfs('A')
dfs_order = dfs('A')

print("BFS:", " -> ".join(bfs_order))
print("DFS:", " -> ".join(dfs_order))


# Draw traversal tree
def draw_tree(order, title):
    G = nx.DiGraph()

    for node in graph:
        for neighbour in graph[node]:
            G.add_edge(node, neighbour)

    pos = {
        'A': (0, 2), 'B': (-1, 1), 'C': (1, 1),
        'D': (-1.5, 0), 'E': (-0.5, 0), 'F': (1, 0)
    }

    labels = {n: f"{n} ({order.index(n)+1})" for n in G.nodes}

    nx.draw(G, pos, with_labels=True, labels=labels,
            node_size=1800, node_color="lightblue",
            font_weight="bold", arrows=False)

    plt.title(title)
    plt.axis("off")
    plt.show()


draw_tree(bfs_order, "BFS Traversal")
draw_tree(dfs_order, "DFS Traversal")
```
